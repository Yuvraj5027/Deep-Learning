# Project 8: Gradient Descent Implementation
A simple Linear Regression model (y = w·x + b) trained using Gradient Descent.

## Task 1: Dataset and Parameter Initialization

```python
import numpy as np
import matplotlib.pyplot as plt

# Simple dataset: hours studied (X) vs marks scored (y)
X = np.array([1, 2, 3, 4, 5])
y = np.array([3, 5, 7, 9, 11])

# Initialize parameters
w = 0.0
b = 0.0

learning_rate = 0.05
iterations = 500

print("Initial w =", w, ", Initial b =", b)
```

## Task 2: Gradient Descent Implementation

```python
n = len(X)
loss_history = []

for i in range(iterations):
    # Prediction
    y_pred = w * X + b

    # Loss (Mean Squared Error)
    loss = np.mean((y - y_pred) ** 2)
    loss_history.append(loss)

    # Gradients
    dw = (2 / n) * np.sum((y_pred - y) * X)
    db = (2 / n) * np.sum(y_pred - y)

    # Update parameters
    w = w - learning_rate * dw
    b = b - learning_rate * db

    if i % 50 == 0:
        print(f"Iteration {i:3d} | w = {w:.4f} | b = {b:.4f} | Loss = {loss:.4f}")

print(f"\nFinal w = {w:.4f}, Final b = {b:.4f}")
```

## Task 3: Loss Reduction Analysis

```python
plt.plot(loss_history)
plt.xlabel("Iteration")
plt.yscale("log")
plt.ylabel("Loss (MSE)")
plt.title("Loss vs Iterations")
plt.grid(True)
plt.show()

print("Initial loss:", round(loss_history[0], 4))
print("Final loss  :", round(loss_history[-1], 6))
```

### Analysis
- The loss starts high because the parameters (w = 0, b = 0) are far from the correct values.
- In every iteration, the parameters are moved in the direction that reduces the error.
- The loss drops sharply in the first few iterations, then decreases slowly and flattens out.
- Final parameters are close to w = 2 and b = 1, which is the true relationship in the data (y = 2x + 1).
